```python
#Using Algorithm 

class Solution:
    def majorityElement(self, nums):
        counter = 0
        v_majorityElement = None
        for num in nums:
            if counter == 0 :
                v_majorityElement = num
            if num == v_majorityElement:
                counter = counter + 1
            else : 
                counter = counter - 1
        
        return v_majorityElement
        
```

```python
Majority = Solution()
Majority.majorityElement([2,2,1,1,1,2,2])
```

```python
#Using sorting 
import math
class Solution:
    def majorityElement(self, nums):
        new_nums = sorted(nums)
        majority = new_nums[math.floor(len(new_nums)/2)]
        return majority
        
```

```python
MajoritySorting = Solution()
MajoritySorting.majorityElement([2,2,1,1,1,2,2])
```

```python
class Solution:
    def singleNumber(self, nums):
        k_dict = {}
        for i in range(0,len(nums)):
            if nums[i] not in k_dict:
                k_dict[nums[i]] = 1
            else:
                k_dict[nums[i]] = k_dict[nums[i]]+1
                
        
        for k, v in k_dict.items():
            if v == 1:
                return k 

        
```

```python
S = Solution()

S.singleNumber([4,1,2,1,2])
```

```python
#Default Dict 
from collections import defaultdict
class Solution:
    def singleNumber(self, nums):
        default_dict = defaultdict(int)
        for num in nums:
            default_dict[num] = default_dict[num] + 1
        
        for num in nums:
            if default_dict[num] ==1:
                return num
    
    
```

```python
S = Solution()

S.singleNumber([4,1,2,1,2])
```

```python
#Using XOR

from collections import defaultdict
class Solution:
    def singleNumber(self, nums):
        result = 0
        for num in nums:
            result = result ^ num
        return result
    
            
```

```python
xor = Solution()

xor.singleNumber([4,1,2,1,2])
```

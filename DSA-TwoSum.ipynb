```python
#Bruitforce method
class Solution(object):
    def twoSum(self, nums, target):
      for x in range(0,len(nums)-1):
            for y in range(x+1, len(nums)):
                if nums[x]+nums[y] == target:
                    return [x,y]
  
S = Solution()
S.twoSum([3,2,4],6)
```

```python

class Solution(object):
    def twoSum(self, nums, target):
        new_list = nums
        new_list = sorted(new_list)
        i = 0
        j = len(new_list)-1
        
        while i<j:
            if new_list[i]+new_list[j] == target:
                if new_list[i] == new_list[j]:
                    return [nums.index(new_list[i]) , nums.index(new_list[j],nums.index(new_list[i])+1 ) ]
                    
                return [nums.index(new_list[i]) , nums.index(new_list[j]) ]
            elif new_list[i]+new_list[j] > target:
                j = j-1
            else: 
                i = i+1
            
```

```python
S = Solution()
S.twoSum([3,3],6)
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        K_dict = {}
        for i in range(0,len(nums)):
                compliment = target - nums[i] 
                if compliment in K_dict.keys():
                    return [K_dict[compliment], i]
                else:
                    K_dict[nums[i]] = i
```

```python
S = Solution()
S.twoSum([2,7,11,15],17)
```
